# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image using OpenCV

img = cv2.imread("baseball.jpg")
```

```python

M = np.float32([[1, 0, 50], [0, 1, 80]])
translated = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))
```

```python
# Display the image using Matplotlib
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB))
plt.title("Translated Image")

plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
rows, cols = img.shape[:2]

M1 = np.float32([[1, 0.5, 0], [0, 1, 0]])
horizontal = cv2.warpAffine(img, M1, (cols, rows))

M2 = np.float32([[1, 0, 0], [0.5, 1, 0]])
vertical = cv2.warpAffine(img, M2, (cols, rows))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontally Sheared Image")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertically Sheared Image")

plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
rows, cols = img.shape[:2]
center = (cols // 2, rows // 2)
```

```python
rotated45 = cv2.warpAffine(img, cv2.getRotationMatrix2D(center, 45, 1), (cols, rows))
```

```python
rotated90 = cv2.warpAffine(img, cv2.getRotationMatrix2D(center, 90, 1), (cols, rows))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(rotated45, cv2.COLOR_BGR2RGB))
plt.title("45° Rotated Image")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(rotated90, cv2.COLOR_BGR2RGB))
plt.title("90° Rotated Image")

plt.show()
```
